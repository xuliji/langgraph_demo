### 延迟执行节点

```python
from IPython.display import Image, display
from loguru import logger
from langgraph.constants import START, END
from langchain_deepseek import ChatDeepSeek
from dotenv import load_dotenv
from langchain_core.messages import SystemMessage, HumanMessage
from typing import NotRequired, Required
from pydantic import Field
from langgraph.graph import MessagesState, StateGraph
from rich import print as rich_print

load_dotenv(override=True)

llm = ChatDeepSeek(model="deepseek-flash")


class CustomerState(MessagesState):
    user_name: NotRequired[str]
    query: NotRequired[str]
    output: NotRequired[str]
    logs: Field(..., default_factory=list)


class Input(MessagesState):
    query: Required[str]


class Output(MessagesState):
    output: Required[str]


def node_1(state: Input) -> CustomerState:
    return {
        "messages": [
            SystemMessage(content="你是一个专业的客服，请用杀生鱼丸的口吻回答我的问题。要俏皮可爱的风格。"),
            HumanMessage(content=state["query"]),
        ]
    }


def query(state: CustomerState) -> CustomerState:
    response = llm.invoke(state["messages"])
    return {
        "messages": [response],
        "output": response.content,
        "logs": [response.additional_kwargs.get("reasoning_content", "")],
    }


def logs(state: CustomerState) -> CustomerState | None:
    """
    打印日志
    :param state: 状态
    :return: None
    """
    print("=====================logs======================")
    rich_print(state["logs"])
    return None


builder = StateGraph(state_schema=CustomerState, input_schema=Input, output_schema=Output)
builder.add_node("node_1", node_1)
# 延迟执行节点
builder.add_node("logs", logs, defer=True)
builder.add_node("query", query)
builder.add_edge(START, "node_1")
builder.add_edge("node_1", "query")
builder.add_edge("query", END)
builder.add_edge(START, "logs")
builder.add_edge("logs", END)

graph = builder.compile()

response = graph.invoke({"query": "你好"})
rich_print(response)
display(Image(graph.get_graph().draw_mermaid_png()))

```
