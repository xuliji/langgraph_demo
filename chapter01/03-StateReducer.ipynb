### 状态Reducer


```python
from typing import Annotated

from langchain_core.messages import BaseMessage
from pydantic import BaseModel, Field


def my_reducer(left: list[str], right: list[str]) -> list[str]:
    # left 是上一个状态，right 是当前状态
    # 返回新的状态
    return left + right


left = ["start", "node_1"]
right = ["node_2", "end"]
reducer = my_reducer(left, right)
print(reducer)
```

```python
from langchain_core.messages import HumanMessage
from langgraph.graph import add_messages
from pydantic import field_validator


class CustomState(BaseModel):
    messages: Annotated[list[BaseMessage], add_messages]
    cur_id: int = Field(..., description="当前节点ID")

    @field_validator("messages")
    @classmethod
    def messages_validator(cls, v: list[BaseMessage]):
        if not v:
            raise ValueError("messages 不能为空")
        messages = [log for log in v if log.content is not None]
        return messages


state = CustomState(messages=[HumanMessage(content="你好", id="2")], cur_id=2122)
state.messages.append(HumanMessage(content="你好", id='1'))
add_messages(state.messages, [HumanMessage(content="你好啊", id='1')])
print(state.messages)
```
